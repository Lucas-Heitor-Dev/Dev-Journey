```python
# In-memory key-value cache demonstration
class SimpleCache:
    def __init__(self):
        self._store = {}
    def set(self, k, v): self._store[k] = v
    def get(self, k, default=None): return self._store.get(k, default)

cache = SimpleCache()
cache.set("session_token", "xyz_auth_9918")
print("Retrieved Cache Token:", cache.get("session_token"))
```
